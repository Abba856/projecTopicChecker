```python
# STEP 1: Import Libraries
import pandas as pd
from sentence_transformers import SentenceTransformer, util
import numpy as np
import nltk
import re
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

# -----------------------------------------------------
# STEP 2: Load Dataset
# -----------------------------------------------------
# Ensure your CSV file (projects.csv) is in the same folder
# The CSV should have at least two columns: 'title' and 'abstract'
df = pd.read_csv(r"paper_details_final_data.csv")

print("✅ Data loaded successfully!")
print(df.head())
```

```python
# -----------------------------------------------------
# STEP 3: Preprocess Text
# -----------------------------------------------------
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

def clean_text(text):
    if pd.isna(text):
        return ""
    text = text.lower()
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    words = text.split()
    words = [lemmatizer.lemmatize(w) for w in words if w not in stop_words]
    return ' '.join(words)

df['clean_title'] = df['final_keywords'].apply(clean_text)
df['clean_abstract'] = df['abstracts'].apply(clean_text)

print("✅ Text preprocessing completed!")
df.head()
```

```python
# -----------------------------------------------------
# STEP 4: Load the Sentence-BERT Model
# -----------------------------------------------------
print("⏳ Loading Sentence-BERT model...")
model = SentenceTransformer('all-MiniLM-L6-v2')
print("✅ Model loaded successfully!")
```

```python
# -----------------------------------------------------
# STEP 5: Encode Titles and Abstracts
# -----------------------------------------------------
title_embeddings = model.encode(df['clean_title'].tolist(), convert_to_tensor=True)
abstract_embeddings = model.encode(df['clean_abstract'].tolist(), convert_to_tensor=True)

print("✅ Text embeddings created!")
```

```python
# -----------------------------------------------------
# STEP 6: Define a Function to Check Similarity
# -----------------------------------------------------
def check_similarity(new_title, new_abstract, top_n=3):
    # Clean input
    new_title_clean = clean_text(new_title)
    new_abstract_clean = clean_text(new_abstract)

    # Encode new inputs
    new_title_emb = model.encode(new_title_clean, convert_to_tensor=True)
    new_abs_emb = model.encode(new_abstract_clean, convert_to_tensor=True)

    # Compute similarity
    title_scores = util.cos_sim(new_title_emb, title_embeddings)[0]
    abs_scores = util.cos_sim(new_abs_emb, abstract_embeddings)[0]

    # Convert to numpy arrays
    title_scores = title_scores.cpu().numpy()
    abs_scores = abs_scores.cpu().numpy()

    # Get top N matches
    top_title_idx = np.argsort(title_scores)[::-1][:top_n]
    top_abs_idx = np.argsort(abs_scores)[::-1][:top_n]

def check_similarity(new_title, new_abstract, top_n=3):
    from sklearn.metrics.pairwise import cosine_similarity
    import numpy as np

    # Encode the new project
    new_title_embedding = model.encode([clean_text(new_title)])
    new_abs_embedding = model.encode([clean_text(new_abstract)])

    # Compute similarity with all existing data
    title_embeddings = model.encode(df['clean_title'].tolist(), show_progress_bar=False)
    abs_embeddings = model.encode(df['clean_abstract'].tolist(), show_progress_bar=False)

    title_scores = cosine_similarity(new_title_embedding, title_embeddings)[0]
    abs_scores = cosine_similarity(new_abs_embedding, abs_embeddings)[0]

    # Get top N most similar projects
    top_title_idx = np.argsort(title_scores)[-top_n:][::-1]
    top_abs_idx = np.argsort(abs_scores)[-top_n:][::-1]

    # --- Display results ---
    print("\n🔹 Top Similar Project Titles:")
    for i in top_title_idx:
        print(f"- {df['final_keywords'][i]}  |  Similarity: {title_scores[i]*100:.2f}%")
        print(f"  Abstract: {df['abstracts'][i][:200]}...\n")

    print("\n🔹 Top Similar Project Abstracts:")
    for i in top_abs_idx:
        print(f"- {df['final_keywords'][i]}  |  Similarity: {abs_scores[i]*100:.2f}%")
        print(f"  Abstract: {df['abstracts'][i][:200]}...\n")
```

```python
#--------------------
# STEP 7: Try It Out!
# -----------------------------------------------------
# Example: Type a new project title and abstract below 👇

new_title = "Design and Development of an Intelligent School Management System"
new_abstract = """
This research focuses on automating school operations using artificial intelligence
to improve record keeping, attendance, and staff management efficiency.
"""

check_similarity(new_title, new_abstract, top_n=3)
```
